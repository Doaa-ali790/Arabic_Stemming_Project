```python
from nltk.stem.isri import ISRIStemmer

isri = ISRIStemmer()

with open("input2.txt", "r", encoding="utf-8") as f:
    words = f.read().split()

print("\n===== ISRI Stemmer Results (Before → After) =====")

with open("output_isri.txt", "w", encoding="utf-8") as out:
    for word in words:
        stem = isri.stem(word)
        line = f"{word}  →  {stem}"
        print(line)
        out.write(line + "\n")

print("✅ Saved in output_isri.txt")
```
