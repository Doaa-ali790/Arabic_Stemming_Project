```python
import re

prefixes = ["ال", "وال", "بال", "كال", "فال", "لل", "و", "ب", "ك", "ف"]
suffixes = ["ها", "ان", "ات", "ون", "ين", "يه", "ية", "ه", "ة", "ت"]

def normalize(word):
    word = re.sub(r'[\u064B-\u0652]', '', word)
    word = re.sub(r'[إأآ]', 'ا', word)
    return word

def kamel_stem(word):
    word = normalize(word)
    for p in prefixes:
        if word.startswith(p) and len(word) > 3:
            word = word[len(p):]
    for s in suffixes:
        if word.endswith(s) and len(word) > 3:
            word = word[:-len(s)]
    return word

with open("input.txt", "r", encoding="utf-8") as f:
    words = f.read().split()

print("\n===== Kamel Stemmer Results (Before → After) =====")

with open("output_kamel.txt", "w", encoding="utf-8") as out:
    for word in words:
        stem = kamel_stem(word)
        line = f"{word}  →  {stem}"
        print(line)
        out.write(line + "\n")

print("✅ Saved in output_kamel.txt")
```
