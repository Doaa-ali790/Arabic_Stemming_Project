```python
from farasa.stemmer import FarasaStemmer

farasa = FarasaStemmer(interactive=False)

with open("input2.txt", "r", encoding="utf-8") as f:
    words = f.read().split()

print("\n===== Farasa Stemmer Results (Before → After) =====")

with open("output_farasa.txt", "w", encoding="utf-8") as out:
    for word in words:
        stem = farasa.stem(word)
        line = f"{word}  →  {stem}"
        print(line)
        out.write(line + "\n")

print("✅ Saved in output_farasa.txt")
```
