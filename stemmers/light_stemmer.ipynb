```python
from nltk.stem.isri import ISRIStemmer

stemmer = ISRIStemmer()

with open("input2.txt", "r", encoding="utf-8") as f:
    words = f.read().split()

print("\n===== Light Stemmer Results (Before → After) =====")

with open("output_light.txt", "w", encoding="utf-8") as out:
    for word in words:
        stem = stemmer.stem(word)
        line = f"{word}  →  {stem}"
        print(line)
        out.write(line + "\n")

print("✅ Saved in output_light.txt")
```
