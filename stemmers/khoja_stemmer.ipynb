```python
import subprocess

jar_path = "khoja-stemmer-command-line.jar"
input_file = "input.txt"
output_file = "output_khoja.txt"

subprocess.run(
    f'java -jar "{jar_path}" "{input_file}" "{output_file}"',
    shell=True
)

print("\n===== Khoja Stemmer Results (Before → After) =====")

with open(input_file, "r", encoding="utf-8") as f1, \
     open(output_file, "r", encoding="utf-8") as f2:

    input_words = f1.read().split()
    output_words = f2.read().split()

    for w, s in zip(input_words, output_words):
        print(f"{w}  →  {s}")

print("✅ Saved in output_khoja.txt")
```
