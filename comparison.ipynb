```python
import pandas as pd
from tabulate import tabulate  # لتنسيق الجدول بشكل جميل في الـ console

# ملفات النتائج لكل خوارزمية
files = {
    "Khoja": "output_khoja.txt",
    "Light": "output_light.txt",
    "ISRI": "output_isri.txt",
    "Kamel": "output_kamel.txt",
    "Farasa": "output_farasa.txt"
}

data = {}

# قراءة الملفات وتجهيز البيانات
for algo, file in files.items():
    with open(file, "r", encoding="utf-8") as f:
        for line in f:
            if "→" in line or "-->" in line:
                # دعم الرمزين → و -->
                if "→" in line:
                    word, stem = line.strip().split("→")
                else:
                    word, stem = line.strip().split("-->")
                word = word.strip()
                stem = stem.strip()
                if word not in data:
                    data[word] = {}
                data[word][algo] = stem

# إنشاء DataFrame
df = pd.DataFrame.from_dict(data, orient="index")
df.index.name = "Original Word"

# عرض الجدول في console بشكل جميل
print("\n===== Comparison Table =====\n")
print(tabulate(df.fillna("-"), headers="keys", tablefmt="fancy_grid", showindex=True))

# حفظ النتائج كملف CSV
df.to_csv("comparison_results.csv", encoding="utf-8-sig")
print("\n✅ Saved as comparison_results.csv")
```
